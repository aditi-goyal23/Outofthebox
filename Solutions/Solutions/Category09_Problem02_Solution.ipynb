# Problem ID: Category09_Problem02 — Longest-Waiting Traffic Signal Agent

### 1. Problem Scenario:
A smart traffic signal controls a two-road intersection with **North-South (NS)** and **East-West (EW)** traffic. Vehicles on each road may have to wait while the other road has a green signal.

The traffic-light agent observes the current waiting time on both roads and decides which direction should receive the next green signal.

The objective is to give green to the direction whose vehicles have been waiting longer, helping to reduce excessive waiting at the intersection.

### 2. System Specifications:
- **State Space ($S$):** The state is represented by `(NS_wait, EW_wait)`, where each value is the current waiting time in seconds.
- **Action Space ($A$):** 2 discrete actions: `0: NS GREEN`, `1: EW GREEN`.
- **Signal Cycle:** Each selected green phase lasts for one simulation cycle.
- **Waiting-Time Rule:** Vehicles on the selected green road have their waiting time reduced. Vehicles on the red road continue to accumulate waiting time.
- **Reward Function:** The reward is the negative of the total waiting time after the signal cycle:
  $$R = -(NS\_wait + EW\_wait)$$
- **Decision Rule:** The agent should select the direction having the larger waiting time. If both waiting times are equal, it should continue the previous green direction.

### 3. Student Implementation Deliverables:
1. **Task 1 (Cell 4):** Implement the `choose_action()` method. Select NS when `NS_wait > EW_wait`, select EW when `EW_wait > NS_wait`, and continue the previous phase when they are equal.
2. **Task 2 (Cell 5):** Implement `update_agent()` to calculate the change in total waiting time and maintain a simple running waiting-time estimate for the selected direction.
3. **Execute & Verify (Cells 6 & 7):** Run the simulation driver and inspect the reward trajectory and average waiting-time trajectory.

```python
# Cell 2: Standard Libraries & Environment Parameters (LOCKED)
import numpy as np
import matplotlib.pyplot as plt
import random

# Set deterministic seed for reproducible results
np.random.seed(42)
random.seed(42)

# Traffic Environment Parameters
INITIAL_NS_WAIT = 4
INITIAL_EW_WAIT = 2

# Waiting-time changes per signal cycle
WAIT_REDUCTION_GREEN = 2
WAIT_INCREASE_RED = 1

# Agent parameter for the running waiting-time estimate
ALPHA = 0.2

# Simulation Parameters
NUM_EPISODES = 100
MAX_STEPS_PER_EPISODE = 30
```

```python
# Cell 3: Environment Simulator Scaffold (LOCKED)
class EnvironmentEngine:
    def __init__(self):
        self.reset()

    def reset(self):
        self.ns_wait = INITIAL_NS_WAIT
        self.ew_wait = INITIAL_EW_WAIT
        self.current_phase = 0  # 0: NS GREEN, 1: EW GREEN
        return self._get_state()

    def _get_state(self):
        return (self.ns_wait, self.ew_wait)

    def step(self, action):
        """Executes one signal cycle and returns (next_state, reward, done)."""
        action = int(action)
        self.current_phase = action

        if action == 0:
            self.ns_wait = max(0, self.ns_wait - WAIT_REDUCTION_GREEN)
            self.ew_wait += WAIT_INCREASE_RED
        else:
            self.ew_wait = max(0, self.ew_wait - WAIT_REDUCTION_GREEN)
            self.ns_wait += WAIT_INCREASE_RED

        next_state = self._get_state()
        reward = -(self.ns_wait + self.ew_wait)

        return next_state, float(reward), False
```

```python
# Cell 4: [TASK 1] Agent Class & Waiting-Time Action Selection (SOLUTION)
class TrafficLightAgent:
    def __init__(self, alpha=ALPHA):
        self.alpha = alpha
        self.current_phase = 0  # 0: NS GREEN, 1: EW GREEN
        self.wait_estimate = np.zeros(2)

    def choose_action(self, state):
        ns_wait, ew_wait = state
        if ns_wait > ew_wait:
            action = 0
        elif ew_wait > ns_wait:
            action = 1
        else:
            action = self.current_phase

        self.current_phase = action
        return action
```

```python
# Cell 5: [TASK 2] Agent Update / Running Waiting-Time Estimate (SOLUTION)
def update_agent(self, state, action, reward, next_state, done):
    ns_wait, ew_wait = state
    next_ns_wait, next_ew_wait = next_state

    total_before = ns_wait + ew_wait
    total_after = next_ns_wait + next_ew_wait
    reduction = max(0, total_before - total_after)

    current_estimate = self.wait_estimate[action]
    self.wait_estimate[action] = (
        current_estimate + self.alpha * (reduction - current_estimate)
    )

# Do not modify this line. It connects the completed function to the agent.
TrafficLightAgent.update_agent = update_agent
```

```python
# Cell 6: Simulation Execution & Test Driver (LOCKED)
env = EnvironmentEngine()
agent = TrafficLightAgent()

rewards_history = []
average_wait_history = []

for ep in range(NUM_EPISODES):
    state = env.reset()
    total_reward = 0.0
    wait_sum = 0.0

    for step in range(MAX_STEPS_PER_EPISODE):
        action = agent.choose_action(state)
        next_state, reward, done = env.step(action)
        agent.update_agent(state, action, reward, next_state, done)

        state = next_state
        total_reward += reward
        wait_sum += sum(state)

    rewards_history.append(total_reward)
    average_wait_history.append(wait_sum / MAX_STEPS_PER_EPISODE)

print(f"Simulation completed across {NUM_EPISODES} episodes.")
print(f"Average Reward (Last 20 Episodes): {np.mean(rewards_history[-20:]):.2f}")
print(f"Average Waiting Time (Last 20 Episodes): {np.mean(average_wait_history[-20:]):.2f}")
print(f"Estimated NS Waiting Reduction: {agent.wait_estimate[0]:.2f}")
print(f"Estimated EW Waiting Reduction: {agent.wait_estimate[1]:.2f}")
```

```python
# Cell 7: Performance Metric Visualization (LOCKED)
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(rewards_history, alpha=0.7)
plt.title("Reward Trajectory per Episode")
plt.xlabel("Episode")
plt.ylabel("Total Reward")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(average_wait_history, alpha=0.7)
plt.title("Average Waiting Time per Episode")
plt.xlabel("Episode")
plt.ylabel("Average Waiting Time")
plt.grid(True)

plt.tight_layout()
plt.show()
```
