### Problem 02: Campus Delivery Robot

**Scenario:** A campus delivery robot must carry a package from the dispatch room to a student services building. Some corridors are blocked by temporary construction, while one cell represents a restricted security zone. The robot learns a route using tabular Q-learning while avoiding blocked and costly cells.

**State space (S):** All 5 × 5 grid cells. State index is `row * 5 + col`, giving 25 discrete states.

**Start state:** `(0, 0)`  
**Goal state:** `(4, 4)`  
**Blocked cells:** `[(1, 2), (2, 2), (3, 1)]`  
**Costly cells:** `[(2, 4)]`

**Action space (A):**
- `0 = UP`
- `1 = DOWN`
- `2 = LEFT`
- `3 = RIGHT`

**Reward / transition mechanics:**
- Reaching the goal gives **+15** and terminates the episode.
- Entering a costly cell gives **-6** and the episode continues.
- A normal valid move gives **-1**.
- Moving into a blocked cell or outside the grid keeps the agent in its current cell and gives **-1**.
- Each episode terminates when the goal is reached or the maximum step limit is reached.
- Restricted security zone (2,4) is traversable but gives a penalty of -6.

**Task:** Use tabular Q-learning to learn an effective navigation policy.

**Student deliverables:**
1. Implement `choose_action(state)` using an epsilon-greedy policy.
2. Implement `update_agent(state, action, reward, next_state, done)` using the Bellman Q-learning update.
3. Run the notebook without changing locked cells and inspect the reward/step curves.

**Core parameters:** `alpha = 0.1`, `gamma = 0.9`, `epsilon = 0.2`.

**Concept focus:** Tabular Q-learning, Bellman Q-table updates, and static/costly-cell avoidance.

```python
class EnvironmentEngine:
    def __init__(self):
        self.rows = ROWS
        self.cols = COLS
        self.start = START
        self.goal = GOAL
        self.obstacles = set(OBSTACLES)
        self.costly_cells = set(COSTLY_CELLS)
        self.state = self.start

    def reset(self):
        self.state = self.start
        return self.state[0] * self.cols + self.state[1]

    def step(self, action):
        row, col = self.state
        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
        drow, dcol = moves[action]
        next_pos = (row + drow, col + dcol)

        if (next_pos[0] < 0 or next_pos[0] >= self.rows or
            next_pos[1] < 0 or next_pos[1] >= self.cols or
            next_pos in self.obstacles):
            next_pos = self.state
            reward = STEP_REWARD
            done = False
        else:
            self.state = next_pos
            if self.state == self.goal:
                reward = GOAL_REWARD
                done = True
            elif self.state in self.costly_cells:
                reward = COSTLY_REWARD
                done = False
            else:
                reward = STEP_REWARD
                done = False

        next_state = self.state[0] * self.cols + self.state[1]
        return next_state, reward, done
```

```python
env = EnvironmentEngine()
agent = QLearningAgent(ROWS * COLS, 4)

episode_rewards = []
episode_steps = []

for episode in range(NUM_EPISODES):
    state = env.reset()
    total_reward = 0
    steps = 0

    for step in range(MAX_STEPS):
        action = agent.choose_action(state)
        next_state, reward, done = env.step(action)
        agent.update_agent(state, action, reward, next_state, done)

        total_reward += reward
        steps += 1
        state = next_state

        if done:
            break

    episode_rewards.append(total_reward)
    episode_steps.append(steps)

print("Training completed.")
print("Episodes:", NUM_EPISODES)
print("Final 20-episode average reward:", round(float(np.mean(episode_rewards[-20:])), 3))
print("Final 20-episode average steps:", round(float(np.mean(episode_steps[-20:])), 3))
```
