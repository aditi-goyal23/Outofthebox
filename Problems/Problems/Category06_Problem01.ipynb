### 1. Problem Scenario:
Two platforms, **Zip** (row player) and **Ola-X** (column player), simultaneously choose a surge policy.
Payoffs are daily profit in ₹ lakh. In each cell the first number is Zip's payoff and the second is Ola-X's.

| Zip \ Ola-X | Low (L) | Cap (C) | Surge (R) |
|---|---|---|---|
| **Aggressive (T)** | 4, 3 | 5, 1 | 6, 2 |
| **Moderate (M)** | 2, 1 | 8, 4 | 3, 6 |
| **Conservative (B)** | 3, 0 | 9, 6 | 2, 8 |

**Questions:** (a) Apply iterated elimination of strictly dominated strategies, showing each step. (b) State the surviving
outcome and verify it is a Nash equilibrium. (c) Does the order of elimination matter? Would the answer change under *weak* dominance?

### 2. System Specifications:
- **Players:** 2 (Zip = rows, Ola-X = columns). **Strategies:** 3 each, indexed 0, 1, 2.
- **Payoff matrices:** `A` (Zip) and `B` (Ola-X), both of shape 3x3.
- **Dominance (pure strategies):** strategy *k* strictly dominates *i* if it is strictly better against **every** remaining opponent strategy;
  *weakly* dominates if it is at least as good everywhere and strictly better somewhere.
- **Equilibrium test:** a cell is a Nash equilibrium if neither player gains by a unilateral deviation.
